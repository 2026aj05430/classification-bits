```python
import pandas as pd
import re
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    roc_auc_score,
    precision_score,
    recall_score,
    f1_score,
    matthews_corrcoef
)

# 2. Load train and test data cleanly using the extracted column names

train_df = pd.read_csv('train_data.csv', sep=';', encoding='utf-8')
test_df = pd.read_csv('test_data.csv', sep=';', encoding='utf-8')
# Clean up any trailing/leading whitespaces from column headers
train_df.columns = train_df.columns.str.strip()
test_df.columns = test_df.columns.str.strip()

# Automatically select the last column as the target
target_column = train_df.columns[-1]
print(f"Target column dynamically set to: '{target_column}'")

# Split features (X) and target (y)
X_train = train_df.drop(columns=[target_column])
y_train = train_df[target_column]

X_test = test_df.drop(columns=[target_column])
y_test = test_df[target_column]

# 3. FEATURE SCALING (StandardScaler)
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# 4. Initialize and train the Logistic Regression model on scaled data
model = LogisticRegression(max_iter=1000, random_state=42)
model.fit(X_train_scaled, y_train)

# 5. Make predictions
y_pred = model.predict(X_test_scaled)
y_prob = model.predict_proba(X_test_scaled) # Probabilities for all classes

# 6. Calculate the required metrics (added multi_class='ovr' and average='weighted' for multi-class support)
accuracy = accuracy_score(y_test, y_pred)
auc_score = roc_auc_score(y_test, y_prob, multi_class='ovr', average='weighted')
precision = precision_score(y_test, y_pred, average='weighted')
recall = recall_score(y_test, y_pred, average='weighted')
f1 = f1_score(y_test, y_pred, average='weighted')
mcc = matthews_corrcoef(y_test, y_pred)

# 7. Print results with clear headings
print("\n==========================================")
print(" LOGISTIC REGRESSION METRICS (WITH SCALING)")
print("==========================================")
print(f"1. Accuracy:  {accuracy:.4f}")
print(f"2. AUC Score: {auc_score:.4f}")
print(f"3. Precision: {precision:.4f}")
print(f"4. Recall:    {recall:.4f}")
print(f"5. F1 Score:  {f1:.4f}")
print(f"6. MCC Score: {mcc:.4f}")
print("==========================================")
```

```python
import pandas as pd
import joblib
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import (
    accuracy_score,
    roc_auc_score,
    precision_score,
    recall_score,
    f1_score,
    matthews_corrcoef
)

# 1. Load the pre-split training and testing datasets
train_df = pd.read_csv('train_data.csv', sep=';', encoding='utf-8')
test_df = pd.read_csv('test_data.csv', sep=';', encoding='utf-8')

# Clean up any trailing/leading whitespaces from column headers just in case
train_df.columns = train_df.columns.str.strip()
test_df.columns = test_df.columns.str.strip()

# 2. Automatically select the last column as the target
target_column = train_df.columns[-1]
print(f"Target column dynamically set to: '{target_column}'")

# Split features (X) and target (y) for both train and test sets
X_train = train_df.drop(columns=[target_column])
y_train = train_df[target_column]

X_test = test_df.drop(columns=[target_column])
y_test = test_df[target_column]

# 3. Initialize and Train the Decision Tree Classifier
# (Decision Trees do not require feature scaling)
model = DecisionTreeClassifier(random_state=42)
model.fit(X_train, y_train)

# 4. Make predictions
y_pred = model.predict(X_test)
y_prob = model.predict_proba(X_test) # Probabilities for AUC score

# 5. Calculate evaluation metrics
accuracy = accuracy_score(y_test, y_pred)
auc_score = roc_auc_score(y_test, y_prob, multi_class='ovr', average='weighted')
precision = precision_score(y_test, y_pred, average='weighted')
recall = recall_score(y_test, y_pred, average='weighted')
f1 = f1_score(y_test, y_pred, average='weighted')
mcc = matthews_corrcoef(y_test, y_pred)

# 6. Print evaluation metrics
print("\n==========================================")
print("   DECISION TREE CLASSIFIER METRICS       ")
print("==========================================")
print(f"1. Accuracy:  {accuracy:.4f}")
print(f"2. AUC Score: {auc_score:.4f}")
print(f"3. Precision: {precision:.4f}")
print(f"4. Recall:    {recall:.4f}")
print(f"5. F1 Score:  {f1:.4f}")
print(f"6. MCC Score: {mcc:.4f}")
print("==========================================")

# 7. Save the trained Decision Tree model to disk
joblib.dump(model, './model/decision_tree_model.pkl')
print("\nDecision Tree model successfully saved as 'decision_tree_model.pkl'")
```

```python
import pandas as pd
import joblib
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import (
    accuracy_score,
    roc_auc_score,
    precision_score,
    recall_score,
    f1_score,
    matthews_corrcoef
)

# 1. Load the pre-split training and testing datasets
train_df = pd.read_csv('train_data.csv', sep=';', encoding='utf-8')
test_df = pd.read_csv('test_data.csv', sep=';', encoding='utf-8')

# Clean up any trailing/leading whitespaces from column headers just in case
train_df.columns = train_df.columns.str.strip()
test_df.columns = test_df.columns.str.strip()

# 2. Automatically select the last column as the target
target_column = train_df.columns[-1]
print(f"Target column dynamically set to: '{target_column}'")

# Split features (X) and target (y) for both train and test sets
X_train = train_df.drop(columns=[target_column])
y_train = train_df[target_column]

X_test = test_df.drop(columns=[target_column])
y_test = test_df[target_column]

# 3. Feature Scaling (Essential for KNN so distance calculations aren't biased by feature magnitudes)
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# 4. Initialize and Train the KNN Classifier (default n_neighbors=5)
model = KNeighborsClassifier(n_neighbors=5)
model.fit(X_train_scaled, y_train)

# 5. Make predictions
y_pred = model.predict(X_test_scaled)
y_prob = model.predict_proba(X_test_scaled) # Probabilities for AUC score

# 6. Calculate evaluation metrics
accuracy = accuracy_score(y_test, y_pred)
auc_score = roc_auc_score(y_test, y_prob, multi_class='ovr', average='weighted')
precision = precision_score(y_test, y_pred, average='weighted')
recall = recall_score(y_test, y_pred, average='weighted')
f1 = f1_score(y_test, y_pred, average='weighted')
mcc = matthews_corrcoef(y_test, y_pred)

# 7. Print evaluation metrics
print("\n==========================================")
print("     K-NEAREST NEIGHBORS (KNN) METRICS     ")
print("==========================================")
print(f"1. Accuracy:  {accuracy:.4f}")
print(f"2. AUC Score: {auc_score:.4f}")
print(f"3. Precision: {precision:.4f}")
print(f"4. Recall:    {recall:.4f}")
print(f"5. F1 Score:  {f1:.4f}")
print(f"6. MCC Score: {mcc:.4f}")
print("==========================================")

# 8. Save the trained KNN model and scaler to disk
joblib.dump(model, './model/knn_model.pkl')
joblib.dump(scaler, './model/knn_scaler.pkl')
print("\nKNN model saved as 'knn_model.pkl' and scaler saved as 'knn_scaler.pkl'")
```

```python
import pandas as pd
import joblib
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.metrics import (
    accuracy_score,
    roc_auc_score,
    precision_score,
    recall_score,
    f1_score,
    matthews_corrcoef
)

# 1. Load the pre-split training and testing datasets
train_df = pd.read_csv('train_data.csv', sep=';', encoding='utf-8')
test_df = pd.read_csv('test_data.csv', sep=';', encoding='utf-8')

# Clean up any trailing/leading whitespaces from column headers just in case
train_df.columns = train_df.columns.str.strip()
test_df.columns = test_df.columns.str.strip()

# 2. Automatically select the last column as the target
target_column = train_df.columns[-1]
print(f"Target column dynamically set to: '{target_column}'")

# Split features (X) and target (y) for both train and test sets
X_train = train_df.drop(columns=[target_column])
y_train = train_df[target_column]

X_test = test_df.drop(columns=[target_column])
y_test = test_df[target_column]

# 3. Feature Scaling (StandardScaler is critical for SVM performance)
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# 4. Initialize and Train the Support Vector Machine Classifier
# (probability=True is required so we can calculate the AUC score using predict_proba)
model = SVC(kernel='rbf', probability=True, random_state=42)
model.fit(X_train_scaled, y_train)

# 5. Make predictions
y_pred = model.predict(X_test_scaled)
y_prob = model.predict_proba(X_test_scaled) # Probabilities for AUC score

# 6. Calculate evaluation metrics
accuracy = accuracy_score(y_test, y_pred)
auc_score = roc_auc_score(y_test, y_prob, multi_class='ovr', average='weighted')
precision = precision_score(y_test, y_pred, average='weighted')
recall = recall_score(y_test, y_pred, average='weighted')
f1 = f1_score(y_test, y_pred, average='weighted')
mcc = matthews_corrcoef(y_test, y_pred)

# 7. Print evaluation metrics
print("\n==========================================")
print(" SUPPORT VECTOR MACHINE (SVM) METRICS     ")
print("==========================================")
print(f"1. Accuracy:  {accuracy:.4f}")
print(f"2. AUC Score: {auc_score:.4f}")
print(f"3. Precision: {precision:.4f}")
print(f"4. Recall:    {recall:.4f}")
print(f"5. F1 Score:  {f1:.4f}")
print(f"6. MCC Score: {mcc:.4f}")
print("==========================================")

# 8. Save the trained SVM model and scaler to disk
joblib.dump(model, './model/svm_model.pkl')
joblib.dump(scaler, './model/svm_scaler.pkl')
print("\nSVM model saved as 'svm_model.pkl' and scaler saved as 'svm_scaler.pkl'")
```
